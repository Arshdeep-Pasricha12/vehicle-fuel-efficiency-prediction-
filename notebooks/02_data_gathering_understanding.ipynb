```python
# ── Imports ──────────────────────────────────────────────
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Plotting style
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette('husl')
pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', '{:.2f}'.format)

print('Libraries loaded ✓')
```

## 2.1 Data Gathering

```python
# Load dataset — '?' in horsepower treated as NaN
df = pd.read_csv('../data/auto-mpg.csv', na_values='?')

# Store a clean copy
df_raw = df.copy()

print(f'Dataset loaded: {df.shape[0]} rows × {df.shape[1]} columns')
df.head(10)
```

## 2.2 Data Understanding

```python
# --- Basic shape & types ---
print('=== Shape ===')
print(df.shape)

print('\n=== Data Types ===')
print(df.dtypes)

print('\n=== Unique Values per Column ===')
print(df.nunique())
```

```python
# --- Statistical Summary ---
print('=== Statistical Summary ===')
df.describe(include='all')
```

```python
# --- Missing Values ---
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)

missing_df = pd.DataFrame({
    'Missing Count': missing,
    'Missing %': missing_pct
}).query('`Missing Count` > 0')

print('=== Missing Values ===')
display(missing_df)
```

```python
# --- Duplicate Check ---
print(f'Duplicate rows: {df.duplicated().sum()}')
```

```python
# --- Value counts for categorical-like columns ---
for col in ['cylinders', 'origin', 'model_year']:
    print(f'\n── {col} value counts ──')
    print(df[col].value_counts().sort_index())
```

```python
# --- Target variable distribution ---
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].hist(df['mpg'], bins=25, color='#4C72B0', edgecolor='white', linewidth=0.5)
axes[0].set_title('Distribution of MPG (Target)', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Miles Per Gallon')
axes[0].set_ylabel('Frequency')

# Box plot
axes[1].boxplot(df['mpg'].dropna(), patch_artist=True,
                boxprops=dict(facecolor='#4C72B0', color='navy'),
                medianprops=dict(color='yellow', linewidth=2))
axes[1].set_title('MPG Box Plot', fontsize=14, fontweight='bold')
axes[1].set_ylabel('Miles Per Gallon')
axes[1].set_xticks([1])
axes[1].set_xticklabels(['MPG'])

plt.tight_layout()
plt.savefig('../plots/01_mpg_distribution.png', dpi=150, bbox_inches='tight')
plt.show()
print(f'Mean MPG: {df["mpg"].mean():.2f} | Median: {df["mpg"].median():.2f} | Std: {df["mpg"].std():.2f}')
```

```python
# Save raw data summary
df.to_csv('../data/auto-mpg.csv', index=False)
print('Data saved ✓')
```
