## 1.1 Business Context

Fuel efficiency (measured in **miles per gallon — MPG**) is one of the most critical factors in the automotive industry. It directly impacts:
- Consumer purchase decisions
- Regulatory compliance (EPA/CAFE standards)
- Fleet management cost optimization
- Environmental sustainability (CO₂ emissions)

For a **telematics and fleet intelligence company like Motorq**, understanding and predicting vehicle fuel efficiency is a core use case — enabling smarter fleet decisions, driver coaching, and sustainability reporting.

---

## 1.2 Problem Statement

> **Given the technical specifications of a vehicle (engine displacement, horsepower, weight, acceleration, model year, and origin), can we accurately predict its fuel efficiency (MPG)?**

This is a **supervised regression problem** where:
- **Input (Features):** `cylinders`, `displacement`, `horsepower`, `weight`, `acceleration`, `model_year`, `origin`
- **Output (Target):** `mpg` (miles per gallon — continuous numeric value)

---

## 1.3 Objectives

1. **Exploratory Analysis** — Understand how vehicle attributes relate to fuel efficiency
2. **Feature Engineering** — Create meaningful features to improve predictive power
3. **Model Building** — Train and compare multiple ML regression models
4. **Model Evaluation** — Select the best-performing model using RMSE, MAE, and R²
5. **Business Insights** — Extract actionable insights for fleet operators

---

## 1.4 Success Criteria

| Metric | Target |
|--------|--------|
| R² Score | > 0.90 |
| RMSE | < 3 MPG |
| MAE | < 2 MPG |

## 1.5 Dataset Overview

| Attribute | Description | Type |
|-----------|-------------|------|
| `mpg` | Miles per gallon (TARGET) | Continuous |
| `cylinders` | Number of engine cylinders | Discrete |
| `displacement` | Engine displacement (cubic inches) | Continuous |
| `horsepower` | Engine horsepower | Continuous |
| `weight` | Vehicle weight (lbs) | Continuous |
| `acceleration` | Time to reach 60 mph (seconds) | Continuous |
| `model_year` | Model year (70-82, i.e., 1970-1982) | Discrete |
| `origin` | Geographic origin (1=USA, 2=Europe, 3=Japan) | Categorical |
| `car_name` | Car name identifier | String |
